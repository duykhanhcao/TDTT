```python
n = int(input())

max_x = 1
max_L = 1

for x in range(1, n + 1):
    so = x          
    do_dai = 1      
    
    while so != 1:
        if so % 2 == 0:
            so = so // 2
        else:
            so = so * 3 + 1
        do_dai += 1
        
    if do_dai > max_L:
        max_L = do_dai
        max_x = x

print(max_x, max_L)
```

```python
a, b= map(int, input().split())
uoc_a=[]
uoc_b=[]
for i in range(1,a+1):
    if a%i==0 and a!=i:
        uoc_a.append(i)
for j in range(1,b+1):
    if b%j==0 and b!=j:
        uoc_b.append(j)
if sum(uoc_a)==b and sum(uoc_b)==a:
    print('true')
else:
    print('false')
    
```
