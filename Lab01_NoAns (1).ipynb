### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math
def triangle_area(a, b, c):
    if a + b <= c or a + c <= b or b + c <= a:
        raise ValueError("Invalid triangle")
    p = (a + b + c) / 2
    area = math.sqrt(p*(p-a)*(p-b)*(p-c))
    return area
print(triangle_area(3, 4, 5))
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def next_letter(x, n):
    if x.islower():
        start = ord('a')
    else:
        start = ord('A')

    result = chr((ord(x) - start + n) % 26 + start)

    return result
print(next_letter('x', 5))
```

### Problem 9
##### Write a function that converts an uppercase letter to lower case. If the input is not a single character, raise ValueError

```python
def to_lowercase(x):
    if len(x) != 1:
        raise ValueError("Input must be one character")

    return x.lower()
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
def triangle(n):
    result = []

    for i in range(1, n + 1):
        result.append('*' * i)

    return '\n'.join(result)


print(triangle(4))
```
