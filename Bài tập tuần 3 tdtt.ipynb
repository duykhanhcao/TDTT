```python
a, b, c, d = map(float, input().split())

if a == b == c == d:
    print('Yes')
elif a != 0:
    cb = b / a
    if b * cb == c and c * cb == d:
        print('Yes')
    else:
        print('No')
else:
    print('No')
```
